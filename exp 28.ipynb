```python
import cv2
import os
import matplotlib.pyplot as plt
from google.colab import files
from ultralytics import YOLO

print("Libraries imported successfully!")
print("OpenCV Version:", cv2.__version__)
uploaded = files.upload()

video_path = list(uploaded.keys())[0]

print("Uploaded video:", video_path)
# Load pre-trained YOLOv8 model
model = YOLO("yolov8n.pt")

print("YOLOv8 model loaded successfully!")
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise ValueError("Unable to open the uploaded video.")

# Get video properties
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

if fps <= 0:
    fps = 25

print("Video FPS:", fps)
print("Video Width:", width)
print("Video Height:", height)
print("Total Frames:", total_frames)
# Vehicle class IDs in the COCO dataset
vehicle_classes = [2, 3, 5, 7]

# Create output video
output_path = "vehicle_detection_output.mp4"

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)

if not out.isOpened():
    cap.release()
    raise ValueError("Unable to create output video.")

frame_count = 0
total_vehicles = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Perform vehicle detection
    results = model.predict(
        source=frame,
        conf=0.25,
        classes=vehicle_classes,
        verbose=False
    )

    # Process detected vehicles
    for result in results:

        for box in result.boxes:

            # Get bounding box coordinates
            x1, y1, x2, y2 = map(
                int, box.xyxy[0].tolist()
            )

            # Get class name and confidence
            class_id = int(box.cls[0])
            confidence = float(box.conf[0])
            label = result.names[class_id]

            # Draw bounding box
            cv2.rectangle(
                frame,
                (x1, y1),
                (x2, y2),
                (0, 255, 0),
                2
            )

            # Display vehicle name and confidence
            text = f"{label}: {confidence:.2f}"

            cv2.putText(
                frame,
                text,
                (x1, max(y1 - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

            total_vehicles += 1

    # Write processed frame into output video
    out.write(frame)

    frame_count += 1

    if frame_count % 30 == 0:
        print("Processed frames:", frame_count)

# Release video resources
cap.release()
out.release()

print("Vehicle detection completed successfully!")
print("Total frames processed:", frame_count)
print("Total vehicle detections across all frames:", total_vehicles)
from IPython.display import Video, display

display(Video(output_path, embed=True))
files.download(output_path)
```
