```python
# HISTOGRAM ANALYSIS BASED ON COLOR LEVELS USING OPENCV

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get image filename
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

# Check image
if img is None:
    raise ValueError("Image not found")

print("Image uploaded successfully")

# Function to analyze color histogram
def analyze_histogram(image):

    # Define color channels
    colors = ('b', 'g', 'r')
    labels = ('Blue', 'Green', 'Red')

    plt.figure(figsize=(10, 5))

    # Calculate histogram for each color
    for i, color in enumerate(colors):

        hist = cv2.calcHist(
            [image], [i], None, [256], [0, 256]
        )

        plt.plot(hist, color=color, label=labels[i])

    plt.title("Color Histogram Analysis")
    plt.xlabel("Pixel Intensity Levels")
    plt.ylabel("Number of Pixels")
    plt.xlim([0, 256])
    plt.legend()
    plt.grid()

    plt.show()

# Display original image
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()

# Analyze histogram
analyze_histogram(img)

print("Color Histogram Analysis Completed Successfully!")

# Save original image
cv2.imwrite("histogram_input.jpg", img)

# Download image
files.download("histogram_input.jpg")
```
