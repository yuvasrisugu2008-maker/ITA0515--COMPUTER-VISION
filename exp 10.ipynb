```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read the image
img = cv2.imread(filename)

if img is None:
    raise ValueError("Image not found. Please upload a valid image.")

# Rotate the image 90 degrees clockwise
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

# Convert BGR images to RGB for display
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

# Display original and rotated images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(rotated_rgb)
plt.title("90 Degree Clockwise Rotation")
plt.axis("off")

plt.show()

# Save the rotated image
cv2.imwrite("rotated_90_clockwise.jpg", rotated)

print("90 Degree Clockwise Rotation Completed Successfully!")

# Download the output image
files.download("rotated_90_clockwise.jpg")
```
