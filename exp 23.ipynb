```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read the input image
img = cv2.imread(filename)

if img is None:
    raise ValueError("Image not found. Please upload a valid image.")

# Create a 5x5 kernel
kernel = np.ones((5, 5), np.uint8)

# Apply Top Hat operation
# Top Hat = Original Image - Opening Image
tophat = cv2.morphologyEx(img, cv2.MORPH_TOPHAT, kernel)

# Convert BGR images to RGB
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
tophat_rgb = cv2.cvtColor(tophat, cv2.COLOR_BGR2RGB)

# Display original and Top Hat images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(tophat_rgb)
plt.title("Top Hat Image")
plt.axis("off")

plt.tight_layout()
plt.show()

# Save the output image
cv2.imwrite("tophat_image.jpg", tophat)

print("Top Hat operation completed successfully!")

# Download the output image
files.download("tophat_image.jpg")
```
