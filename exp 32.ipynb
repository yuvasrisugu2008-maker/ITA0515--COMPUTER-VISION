```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

def create_colored_boxes():
    # Get image size from user
    height = int(input("Enter image height: "))
    width = int(input("Enter image width: "))

    if height <= 0 or width <= 0:
        print("Please enter positive dimensions.")
        return

    # Create white image
    image = np.ones((height, width, 3), dtype=np.uint8) * 255

    # Calculate box size (1/10th of image)
    box_h = max(1, height // 10)
    box_w = max(1, width // 10)

    # Draw boxes at four corners
    # OpenCV uses BGR color format

    # Top-left: Black
    image[0:box_h, 0:box_w] = (0, 0, 0)

    # Top-right: Blue
    image[0:box_h, width-box_w:width] = (255, 0, 0)

    # Bottom-left: Green
    image[height-box_h:height, 0:box_w] = (0, 255, 0)

    # Bottom-right: Red
    image[height-box_h:height, width-box_w:width] = (0, 0, 255)

    # Display image
    plt.figure(figsize=(7, 5))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title("White Image with Four Colored Boxes")
    plt.axis("off")
    plt.show()

    # Save image
    cv2.imwrite("colored_boxes.jpg", image)
    print("Image saved as colored_boxes.jpg")

create_colored_boxes()
```
