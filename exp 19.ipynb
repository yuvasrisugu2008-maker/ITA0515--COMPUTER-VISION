```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read the input image
img = cv2.imread(filename)

if img is None:
    raise ValueError("Image not found. Please upload a valid image.")

# Create a kernel
kernel = np.ones((5, 5), np.uint8)

# Apply erosion operation
erosion = cv2.erode(img, kernel, iterations=1)

# Convert BGR images to RGB for display
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
erosion_rgb = cv2.cvtColor(erosion, cv2.COLOR_BGR2RGB)

# Display original and eroded images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(erosion_rgb)
plt.title("Eroded Image")
plt.axis("off")

plt.tight_layout()
plt.show()

# Save the eroded image
cv2.imwrite("eroded_image.jpg", erosion)

print("Erosion operation completed successfully!")

# Download the output image
files.download("eroded_image.jpg")
```
