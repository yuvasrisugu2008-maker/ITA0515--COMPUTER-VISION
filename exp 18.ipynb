```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read the source image
img = cv2.imread(filename)

if img is None:
    raise ValueError("Image not found. Please upload a valid image.")

# Get image dimensions
height, width = img.shape[:2]

# Select ROI coordinates
x1 = width // 4
y1 = height // 4
x2 = 3 * width // 4
y2 = 3 * height // 4

# Crop the Region of Interest (ROI)
roi = img[y1:y2, x1:x2]

# Copy the original image
copied_img = img.copy()

# Paste the cropped ROI at a different position
roi_height, roi_width = roi.shape[:2]

# Destination position (top-left corner)
paste_x = 10
paste_y = 10

# Paste ROI into the copied image
copied_img[
    paste_y:paste_y + roi_height,
    paste_x:paste_x + roi_width
] = roi

# Convert images from BGR to RGB
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)
pasted_rgb = cv2.cvtColor(copied_img, cv2.COLOR_BGR2RGB)

# Display original, cropped and pasted images
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(roi_rgb)
plt.title("Cropped ROI")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(pasted_rgb)
plt.title("Copied and Pasted Image")
plt.axis("off")

plt.tight_layout()
plt.show()

# Save output images
cv2.imwrite("cropped_roi.jpg", roi)
cv2.imwrite("pasted_image.jpg", copied_img)

print("Image Cropping, Copying and Pasting Completed Successfully!")

# Download output images
files.download("cropped_roi.jpg")
files.download("pasted_image.jpg")
```
