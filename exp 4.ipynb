```python
# HISTOGRAM EQUALIZATION USING OPENCV

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get image filename
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

# Check image
if img is None:
    raise ValueError("Image not found")

print("Image uploaded successfully")

# Convert image into grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Histogram Equalization
equalized = cv2.equalizeHist(gray)

# Display original and equalized images
plt.figure(figsize=(14, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(equalized, cmap="gray")
plt.title("Histogram Equalized Image")
plt.axis("off")

# Display original histogram
plt.subplot(2, 2, 3)
plt.hist(gray.ravel(), 256, [0, 256], color="blue")
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

# Display equalized histogram
plt.subplot(2, 2, 4)
plt.hist(equalized.ravel(), 256, [0, 256], color="red")
plt.title("Equalized Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

# Save equalized image
cv2.imwrite("histogram_equalized.jpg", equalized)

print("Histogram Equalization completed successfully!")

# Download output image
files.download("histogram_equalized.jpg")
```
