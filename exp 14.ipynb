```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read the image
img = cv2.imread(filename)

if img is None:
    raise ValueError("Image not found. Please upload a valid image.")

# Get image dimensions
height, width = img.shape[:2]

# Define four points in the original image
pts1 = np.float32([
    [0, 0],
    [width - 1, 0],
    [0, height - 1],
    [width - 1, height - 1]
])

# Define four points in the output image
pts2 = np.float32([
    [width * 0.1, height * 0.1],
    [width * 0.9, height * 0.2],
    [width * 0.2, height * 0.9],
    [width * 0.8, height * 0.8]
])

# Calculate perspective transformation matrix
matrix = cv2.getPerspectiveTransform(pts1, pts2)

# Apply perspective transformation
transformed = cv2.warpPerspective(img, matrix, (width, height))

# Convert BGR to RGB for displaying
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
transformed_rgb = cv2.cvtColor(transformed, cv2.COLOR_BGR2RGB)

# Display original and transformed images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(transformed_rgb)
plt.title("Perspective Transformed Image")
plt.axis("off")

plt.show()

# Save the transformed image
cv2.imwrite("perspective_transformed.jpg", transformed)

print("Perspective Transformation Completed Successfully!")

# Download output image
files.download("perspective_transformed.jpg")
```
