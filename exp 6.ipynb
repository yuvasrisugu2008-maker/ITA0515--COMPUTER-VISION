```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read the image
img = cv2.imread(filename)

if img is None:
    raise ValueError("Image not found. Please upload a valid image.")

# Convert the image into grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert grayscale image into binary image
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Create a kernel
kernel = np.ones((5, 5), np.uint8)

# Apply erosion operation
erosion = cv2.erode(binary, kernel, iterations=1)

# Display original image
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Display binary image
plt.subplot(1, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

# Display eroded image
plt.subplot(1, 3, 3)
plt.imshow(erosion, cmap="gray")
plt.title("Eroded Image")
plt.axis("off")

plt.show()

# Save the output image
cv2.imwrite("eroded_output.jpg", erosion)

print("Image Erosion Completed Successfully!")

# Download the output
files.download("eroded_output.jpg")
```
