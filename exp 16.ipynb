```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read the image
img = cv2.imread(filename)

if img is None:
    raise ValueError("Image not found. Please upload a valid image.")

# Convert image into grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Gaussian Blur to reduce noise
blur = cv2.GaussianBlur(gray, (3, 3), 0)

# Apply Sobel filter in X direction
sobel_x = cv2.Sobel(
    blur, cv2.CV_64F, 1, 0, ksize=3
)

# Apply Sobel filter in Y direction
sobel_y = cv2.Sobel(
    blur, cv2.CV_64F, 0, 1, ksize=3
)

# Convert results into absolute values
sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

# Combine X and Y direction edges
sobel_combined = cv2.addWeighted(
    sobel_x, 0.5, sobel_y, 0.5, 0
)

# Display original and filtered images
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X Direction")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y Direction")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(sobel_combined, cmap="gray")
plt.title("Combined Sobel Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.show()

# Save the output image
cv2.imwrite("sobel_output.jpg", sobel_combined)

print("Sobel Edge Detection Completed Successfully!")

# Download the output image
files.download("sobel_output.jpg")
```
