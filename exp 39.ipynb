```python
import cv2
import os
from google.colab import files
from IPython.display import Video, display

import cv2
import os
from google.colab import files
from IPython.display import Video, display

def reverse_slow_motion():

    # Upload video
    uploaded = files.upload()
    filename = next(iter(uploaded))

    # Open video
    cap = cv2.VideoCapture(filename)

    if not cap.isOpened():
        raise ValueError("Unable to open the video.")

    # Get video properties
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    if fps <= 0 or width <= 0 or height <= 0:
        cap.release()
        raise ValueError("Invalid video properties.")

    # Read all frames
    frames = []

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        frames.append(frame)

    cap.release()

    if len(frames) == 0:
        raise ValueError("No frames found in the video.")

    # Reverse frame order
    frames.reverse()

    # Create output video
    output_file = "reverse_slow_motion.mp4"

    # Half-speed playback
    slow_fps = max(1, fps / 2)

    fourcc = cv2.VideoWriter_fourcc(*"mp4v")

    out = cv2.VideoWriter(
        output_file,
        fourcc,
        slow_fps,
        (width, height)
    )

    if not out.isOpened():
        raise RuntimeError("Unable to create output video.")

    # Write reversed frames
    for frame in frames:
        out.write(frame)

    out.release()

    print("Reverse slow-motion video created successfully!")

    # Display output video
    display(Video(output_file, embed=True))

    # Download output
    files.download(output_file)

reverse_slow_motion()
```
